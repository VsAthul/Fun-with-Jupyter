```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_csv('dataset_for_stroke_prediction.csv')
data.head()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data['bmi'].replace(np.nan, data['bmi'].mean(), inplace=True)
```

```python
data.info()
```

```python
data.head()
```

```python
category_names = data.select_dtypes('object').columns.tolist()
le = LabelEncoder()
for x in category_names:
  data[x]= le.fit_transform(data[x])
data
```

```python
data.corr()['stroke']
```

```python
features = data[["age","hypertension","heart_disease","avg_glucose_level"]]
features
```

```python
x=features
y=data['stroke']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size = 0.2, random_state=40)
```

```python
model = DecisionTreeClassifier(criterion='entropy', max_depth=4)
model.fit(xtrain,ytrain)
pred = model.predict(xtest)
print("Accuracy : ", accuracy_score(ytest, pred))
```
